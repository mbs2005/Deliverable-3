```python
import random

class Dice:
    "A class where you can make a dice with a number of given sides"
    
    def __init__(self, sides: int = 6) -> None:
        "Makes a standard dice with 6 sides"
        self.sides = sides
        
    def roll(self) -> int:
        "Roll the dice and return a random number between 1 and the number of sides"
        return random.randint(1, self.sides)
    
    def __repr__(self) -> str:
        "Return a string representation of the dice object"
        return f"Dice(sides={self.sides})"
    
    def __call__(self) -> int:
        "Allows a dice instance to be called directly like a function to roll it"
        return self.roll()

"""# Try to use the dice function
terning = Dice()
print(terning.roll())
print(terning)

d20 = Dice(sides = 20)
print(d20())"""
```

```python
from dice import Dice

class Weapon:
    "A class representing a weapon that uses the dice for damage calculations"
    
    def __init__(self, name: str, damage_dice: Dice) -> None:
        "Initializes a weapon with a name and a damage-dealing dice"
        self.name: str = name
        self.damage_dice = damage_dice
        
    def roll_damage(self) -> int:
        "Rolls the weapon damage dice and return the resulting damage"
        return self.damage_dice.roll()
    
    def __repr__(self) -> str:
        "Return a string representation of the weapon object"
        return f"Weapon(name = '{self.name}', damage_dice = {self.damage_dice})"
```
